## 1. Introduction

Hierarchical clustering is a machine learning technique used to arrange data into different groups according to their similarities. It creates clusters in a tree-like structure called a hierarchy. This method has two types: agglomerative clustering, which combines individual data points into larger groups, and divisive clustering, which separates large groups into smaller clusters. A dendrogram is used to represent the clustering structure visually.

## 2. Importing required libraries

I import the libraries needed to create sample data, calculate distances, cluster points and plot graphs.

```python

import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import linkage, dendrogram
from scipy.spatial.distance import pdist, squareform
from sklearn.cluster import AgglomerativeClustering        #AP24110010828
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler
```

```python

X, _ = make_blobs(n_samples=15, centers=3, cluster_std=0.8, random_state=42) #AP24110010828
print("Dataset shape:", X.shape)
print(X[:5])
```

## 9. Divisive clustering (simple DIANA-style demonstration)

Divisive clustering begins with all points in one cluster and splits a cluster into smaller groups. The article’s custom DIANA code appears to contain typing errors, so this cell is a simple educational implementation: it repeatedly splits the cluster with the largest spread using a 2-way split. It demonstrates the splitting idea, rather than reproducing the article’s exact code.

```python

from sklearn.cluster import KMeans

def simple_divisive(X, n_clusters=3):
    clusters = [np.arange(len(X))]
    while len(clusters) < n_clusters:
        # Split the group with the largest spread
        spreads = [np.sum(np.var(X[idx], axis=0)) for idx in clusters]
        split_idx = int(np.argmax(spreads))
        group = clusters.pop(split_idx)
        if len(group) < 2:                     #AP24110010828
            clusters.append(group)
            break
        splitter = KMeans(n_clusters=2, random_state=42, n_init=10)
        sublabels = splitter.fit_predict(X[group])
        left = group[sublabels == 0]
        right = group[sublabels == 1]
        if len(left) == 0 or len(right) == 0:
            clusters.append(group)
            break
        clusters.extend([left, right])         #AP24110010828
    final_labels = np.zeros(len(X), dtype=int)
    for cluster_id, idx in enumerate(clusters):
        final_labels[idx] = cluster_id
    return final_labels

div_labels = simple_divisive(X_scaled, n_clusters=3)
print("Divisive cluster labels:", div_labels)
```

## 10. Visualizing divisive clusters

This plot shows the three groups produced by the simple splitting demonstration.

```python
                                             #AP24110010828
plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=div_labels)
plt.title("Divisive Clustering - Simple Demonstration")
plt.xlabel("Scaled Feature 1")
plt.ylabel("Scaled Feature 2")
plt.show()
```

## 11. Conclusion

Hierarchical clustering groups data points into clusters based on how similar they are to each other. It follows two approaches: agglomerative clustering, which combines smaller groups into bigger clusters, and divisive clustering, which divides a large group into smaller ones. A dendrogram is used to show how the clusters are formed. This technique is useful for understanding data patterns and relationships, especially in small datasets. However, it may require more time and computational resources when working with large datasets.
