```python
#Line plot with markers and grid
#AP24110010828
import matplotlib.pyplot as plt

# Define X and Y data points
X = [12, 34, 23, 45, 67, 89]
Y = [1, 3, 67, 78, 7, 5]

# Plot the graph
plt.plot(X, Y, marker='o', markerfacecolor='red')

# Axis labels
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
#AP24110010828
# Add gridlines
plt.grid(color='green', linestyle='--', linewidth=0.5)

# Add title
plt.title(
    "Number of Users of a Particular Language",
    fontweight='bold',
    pad=10
)

# Display the plot
plt.show()
```

```python
#Multiple lines
#AP24110010828
import matplotlib.pyplot as plt
import numpy as np

# Create data
x = [1, 2, 3, 4, 5]
y = [3, 3, 3, 3, 3]

# Plot lines
plt.plot(x, y, label="Line 1", linestyle="-") #AP24110010828
plt.plot(y, x, label="Line 2", linestyle="--")
plt.plot(x, np.sin(x), label="Curve 1", linestyle="-.")
plt.plot(x, np.cos(x), label="Curve 2", linestyle=":")

# Display legend and graph
plt.legend()
plt.show()
```

```python
#Stacked bar chart

import matplotlib.pyplot as plt
import pandas as pd

# Create dataset
data = [
    ['A', 10, 20, 10, 26], #AP24110010828
    ['B', 20, 25, 15, 21],
    ['C', 12, 15, 19, 6],
    ['D', 10, 18, 11, 19]
]

# Create DataFrame
df = pd.DataFrame(
    data,
    columns=['Team', 'Round 1', 'Round 2', 'Round 3', 'Round 4']
)

# Display data
print(df)

# Draw stacked bar chart
df.plot( #AP24110010828
    x='Team',
    kind='bar',
    stacked=True,
    title='Stacked Bar Graph by DataFrame'
)

plt.show()
```

```python
#Two histograms
#Ap24110010828

import matplotlib.pyplot as plt

# First age group
age_g1 = [
    1, 3, 5, 10, 15, 17, 18, 16, 19, 21,
    23, 28, 30, 31, 33, 38, 32, 40, 45, 43,
    49, 55, 53, 63, 66, 85, 80, 57, 75, 93, 95
]

# Second age group
age_g2 = [
    6, 4, 15, 17, 19, 21, 28, 23, 31, 36,
    39, 32, 50, 56, 59, 74, 79, 34, 98, 97,
    95, 67, 69, 92, 45, 55, 77, 76, 85
]

# Plot first histogram
plt.hist(
    age_g1,
    label='Age Group 1',
    bins=14, #AP24110010828
    edgecolor='red',
    alpha=0.5
)

# Plot second histogram
plt.hist(
    age_g2,
    label='Age Group 2',
    bins=14,
    edgecolor='orange',
    alpha=0.5
)

# Display legend and graph
plt.legend()
plt.show()
```

```python
#Two scatter datasets
#AP24110010828

import matplotlib.pyplot as plt
from matplotlib import style

# Apply style
style.use('ggplot')

# First dataset
x = [1, 1.5, 2, 2.5, 3, 3.5, 3.6]
y = [7.5, 8, 8.5, 9, 9.5, 10, 10.5]

# Second dataset
x1 = [8, 8.5, 9, 9.5, 10, 10.5, 11]
y1 = [3, 3.5, 3.7, 4, 4.5, 5, 5.2]

# First scatter plot
plt.scatter(
    x, y,
    label='High Income Low Saving',
    color='red'
)

# Second scatter plot    #AP24110010828
plt.scatter(
    x1, y1,
    label='Low Income High Savings',
    color='blue'
)

# Axis labels and title
plt.xlabel('Saving * 100')
plt.ylabel('Income * 1000')
plt.title('Scatter Plot')

# Legend and display
plt.legend()
plt.show()
```

```python
#Pie chart
#AP24110010828
import matplotlib.pyplot as plt

# Values for each slice
slices = [7, 2, 2, 13]

# Labels for each slice
activities = ['Sleeping', 'Eating', 'Working', 'Playing']

# Colors for slices
cols = ['cyan', 'magenta', 'red', 'blue']

# Draw pie chart
plt.pie(
    slices,   #AP24110010828
    labels=activities,
    colors=cols,
    autopct='%1.1f%%'
)

# Chart title
plt.title('Pie Plot')

# Display chart
plt.show()
```

```python
#Seaborn line plot
#AP24110010828
import seaborn as sns
import matplotlib.pyplot as plt

# Load Iris dataset
data = sns.load_dataset("iris")

# Draw line plot
sns.lineplot(
    x="sepal_length",
    y="sepal_width",
    data=data
)  #AP24110010828

# Set X-axis limits
plt.xlim(left=5)

# Display plot
plt.show()
```

```python
#Seaborn with Matplotlib
#AP24110010828
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Generate a 10 x 10 matrix of random numbers
data = np.random.randint(
    low=1,
    high=100,
    size=(10, 10)
)

# Print the data
print("The data to be plotted:\n")
print(data)

# Plot the heatmap
hm = sns.heatmap(
    data=data,
    annot=True
) #AP24110010828

# Display heatmap
plt.show()
```

```python
#Basic Heatmap
#AP24110010828
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Generate data
data = np.random.randint(1, 100, size=(10, 10))

# Create heatmap
sns.heatmap(
    data=data,
    annot=True,
    fmt='d',
    cmap='coolwarm',
    linewidths=0.5,
    linecolor='white',
    cbar=True
)
 #AP24110010828
plt.title("Basic Heatmap in Python")
plt.show()
```
