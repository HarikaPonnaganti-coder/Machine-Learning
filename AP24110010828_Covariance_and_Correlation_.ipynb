```python
import pandas as pd
data = {
    'Study_Hours': [2, 4, 6, 8, 10],
    'Marks': [20, 40, 50, 70, 90]
}
df = pd.DataFrame(data)       #AP24110010828
covariance = df['Study_Hours'].cov(df['Marks'])
correlation = df['Study_Hours'].corr(df['Marks'])  #AP24110010828


print("Covariance:", covariance)
print("Correlation:", correlation)
```

Compute Covariance and Correlation Between a Pair of Attributes

program uses the Pandas library to calculate covariance and correlation between two attributes. First, it imports Pandas and creates a dataset containing two attributes, such as study hours and marks. The dataset is converted into a DataFrame using pd.DataFrame(). The .cov() function calculates covariance to determine how the two attributes vary together, while the .corr() function calculates the Pearson correlation coefficient to measure the strength and direction of their linear relationship. Finally, the print() function displays the covariance and correlation values.   #AP24110010828

```python

 import pandas as pd       #Covariance Matrix and Correlation Matrix
data = {
    'A': [2, 4, 6, 8, 10],
    'B': [1, 3, 5, 7, 9],
    'C': [5, 10, 15, 20, 25] #AP24110010828
}

df = pd.DataFrame(data)
print("Covariance Matrix:")
print(df.cov())

# AP24110010828
print("\nCorrelation Matrix:")
print(df.corr())

```

Covariance Matrix and Correlation Matrix

program uses Pandas to calculate the covariance matrix and correlation matrix for multiple attributes. First, it imports Pandas and creates a dataset containing attributes such as A, B, and C. The dataset is converted into a DataFrame using pd.DataFrame(). The df.cov() function calculates the covariance matrix, showing how each pair of attributes varies together, while the df.corr() function calculates the correlation matrix, indicating the strength and direction of the linear relationship between each pair of attributes. Finally, both matrices are displayed using the print() function. #AP24110010828
